```python
import pandas as pd
import requests
import json
import os


# Load Bayern Cloud API key from environment variables
BAYERN_CLOUD_API_KEY = os.getenv('BAYERN_CLOUD_API_KEY')

# We are not using 'parkplatz-fredenbruecke-1' and 'skiwanderzentrum-zwieslerwaldhaus-2' because of inconsistency in sending data to the cloud
pois = {
    "Schwellhäusl": "03014f93-fba2-4d35-8375-3b481b8dcf0b",
    "Falkensteinschutzhaus": "e07cb08d-fd57-4425-b111-9f323358572e",
    "Waldschmidthaus": "84c6ea24-33c0-46ef-a999-123aa364595d",
    "Racheldiensthütte": "cbb03546-b70d-40c7-95ed-97e450736637",
    "Lusenschutzhaus": "9e57c2d1-47f6-421d-90a4-357afd996fe6",
    "Nationalpark Café Wistlberg": "8dea49d0-3319-490a-ad97-8b53dc062247",
    "HEH": "5d2daf54-6ec6-41f5-ba66-6548913560cf",
    "WGM": "2e431d94-0071-4d40-9d52-1d3d023ad72a",
    "HZW": "cd83f3cf-f169-4d1e-a118-2b3629d47ba6",
}


def get_bayern_cloud_poi(
    poi_name: str,
    poi_id: str
):
    API_endpoint = f'https://data.bayerncloud.digital/api/v4/endpoints/list_poi/{poi_id}'

    request_params = {
        'token': BAYERN_CLOUD_API_KEY
    }

    response = requests.get(API_endpoint, params=request_params)
    response_json = response.json()

    try:
        print(response_json["@graph"][0]["openingHoursSpecification"])
    except:
        print(f"{poi_name} has no opening hours")
        print(response_json)

for key, value in pois.items():
    get_bayern_cloud_poi(poi_name=key, poi_id=value)
```
